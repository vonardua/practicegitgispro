```python
coordinates = [
    (40.7128, -74.0060),   # New York City, USA
    (30.0444, 31.2357),    # Cairo, Egypt
    (35.6762, 139.6503),   # Tokyo, Japan
    (-33.9249, 18.4241),   # Cape Town, South Africa
    (-6.2088, 106.8456),   # Jakarta, Indonesia
    (-34.6037, -58.3816),  # Buenos Aires, Argentina
    (-33.8688, 151.2093),  # Sydney, Australia
    (51.9225, 4.4792)      # Rotterdam, Netherlands
]

for lat, lon in coordinates:
    if lat > 0:
        print(f"This city is in the Northern Hemisphere: {lat}, {lon}")
    else:
        print(f"This city is in the Southern Hemisphere: {lat}, {lon}")
```

```python
for lat,lon in coordinates:
    print(lat)
```

```python
counter = 0
while counter < len(coordinates):
    lat, lon = coordinates[counter]
    if lat > 0:
        print(f"City in the Northern Hemisphere {coordinates[counter]}")
    else:
        break
    counter += 1
```

```python
for lat, lon in coordinates:
    if lon > 0:
        print(f"({lat}, {lon}) is a city in the Eastern Hemisphere")
    else:
        print(f"({lat},{lon}) is a city in the Western Hemisphere")
```

```python
counter = 0
southern_counter = 0
northern_counter = 0
eastern_counter = 0
western_counter = 0

for lat, lon in coordinates:
    if lat < 0:
        northern_counter += 1
    else:
        southern_counter += 1

    if lon < 0:
        western_counter += 1
    else:
        eastern_counter += 1

print(f"There are {northern_counter} cities in the Northern Hemisphere")
```

```python
import random

target_latitude = 50
target_longitude = 50
attempts = 0
max_attempts = 10

print(f"Searching for coordinates where latitude is more than {target_latitude} and longitude is more than {target_longitude}")

while attempts <= max_attempts:
    random_lat = random.uniform(-90,90)
    random_lon = random.uniform(-180, 180)
    attempts += 1

    print(f"Attempt no. {attempts} found coordinate ({random_lat:.4f}, {random_lon:.4f})")
    if random_lat >= target_latitude and random_lon >= target_longitude:
        print (f"Successfully got coordinate where latitude > {target_latitude} and longitude is > {target_longitude}")
        break
    else:
        print(f"Unsuccessfully got coordinate where latitude > {target_latitude} and longitude is > {target_longitude}")
```
