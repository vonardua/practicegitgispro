```python
# Exercise 5 Nested data structures
city_info = {
    "Tokyo": {
        "population": 13515271,
        "coordinates": (35.6895, 139.6917)
    }
}

tokyo_info = city_info["Tokyo"]
print(f"Tokyo's population is {tokyo_info['population']:,}")
print(f"Tokyo's coordinates are {tokyo_info['coordinates']}")

tokyo_info["population"] = 14000000

print("Tokyo info:", city_info)
```
