```python
river = "Amazon River"
river_lc = river.lower()
river_uc = river.upper()

print(f"in lowercase: {river_lc}")
print(f"in uppercase: {river_uc}")

river_country = "Brazil"
river_full = river + ", " + river_country

print(f"Location: {river_full}")

print((river_full + " - " )* 3)
# Exercise 1
```

```python
location = "40.7128N, 74.0060W"
location_parts = location.split(", ")
print(location_parts)

lat, lon = location_parts
lat_num = lat[:-1]
lon_num = lon[:-1]

lat_float = float(lat_num)
lon_float = float(lon_num)

print(lat_float, lon_float)
# this is exercise 2
```

```python
table_name = "cities"
condition = "population > 100000"
min_elevation = 100

sql_query = f"""SELECT
FROM {table_name} 
WHERE {condition}
AND elevation is > {min_elevation} meters"""

print(sql_query)
# this is exercise 3
```
